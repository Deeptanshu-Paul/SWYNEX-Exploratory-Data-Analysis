```python
duplicate_counts = [
    train.duplicated().sum(),
    len(train.drop_duplicates())
]

labels = ['Duplicate Rows', 'Unique Rows']

plt.figure(figsize=(8,5))

plt.bar(
    ['Duplicate Rows', 'Unique Rows'],
    [train.duplicated().sum(), len(train.drop_duplicates())]
)

plt.title('Duplicate vs Unique Records')
plt.ylabel('Number of Rows')

plt.tight_layout()
plt.savefig('duplicate_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
disease_counts = train['prognosis'].value_counts()

print(disease_counts)
```

```python
plt.figure(figsize=(14,10))

disease_counts.sort_values().plot(kind='barh')

plt.title('Disease Class Distribution')
plt.xlabel('Number of Records')
plt.ylabel('Disease')

plt.tight_layout()
plt.savefig('disease_class_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
symptom_columns = train.drop(columns=['prognosis']).columns

symptom_frequency = train[symptom_columns].sum().sort_values(
    ascending=False
)

print(symptom_frequency.head(20))
```

```python
plt.figure(figsize=(12,8))

symptom_frequency.head(20).sort_values().plot(kind='barh')

plt.title('Top 20 Most Frequent Symptoms')
plt.xlabel('Number of Records')
plt.ylabel('Symptom')

plt.tight_layout()
plt.savefig('symptom_frequency.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
train['symptom_count'] = train[symptom_columns].sum(axis=1)

print(train['symptom_count'].describe())
```

```python
plt.figure(figsize=(10,6))

plt.hist(train['symptom_count'], bins=15)

plt.title('Distribution of Number of Symptoms per Record')
plt.xlabel('Number of Symptoms')
plt.ylabel('Frequency')

plt.tight_layout()
plt.savefig('symptom_count_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
combination_counts = (
    train[symptom_columns]
    .astype(str)
    .agg(''.join, axis=1)
    .value_counts()
)

print(combination_counts.head(10))
```

```python
top_symptoms = symptom_frequency.head(20).index

corr = train[top_symptoms].corr()
```

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Between Frequently Occurring Symptoms')
plt.tight_layout()
plt.show()
```

```python
train_frequency = train[symptom_columns].mean()
test_frequency = test[symptom_columns].mean()

comparison = pd.DataFrame({
    'Training': train_frequency,
    'Testing': test_frequency
})

comparison['Difference'] = (
    comparison['Testing'] - comparison['Training']
)

comparison.sort_values(
    'Difference',
    key=abs,
    ascending=False
).head(20)
```

```python
# Remove duplicate records based on symptom patterns
train_unique = train.drop_duplicates(
    subset=symptom_columns
)

# Add prognosis separately using a normal Python list
merge_columns = list(symptom_columns) + ['prognosis']

# Compare testing patterns with training patterns
test_check = test.merge(
    train_unique[merge_columns],
    on=list(symptom_columns),
    how='left',
    suffixes=('_test', '_train')
)

# Count testing patterns that are also found in training
matched_patterns = test_check['prognosis_train'].notna().sum()

print("Testing patterns found in training:", matched_patterns)
print("Total testing records:", len(test))
```
